### 'NA' Values

```python
data={'one':pd.Series([1,2,5],index=['a','b','e']),'two':pd.Series([1,2,3,4],index=['a','b','c','d'])}
table=pd.DataFrame(data)
table
```

```python
imputer=SimpleImputer(missing_values=np.nan,strategy='mean')
imputer
imputer.fit(table)
data_trans=imputer.transform(table)
data_trans=pd.DataFrame(data_trans,columns=['one','two'])
data_trans
```

```python
imputer=SimpleImputer(missing_values=np.nan,strategy='median')
imputer
imputer.fit(table)
data_trans=imputer.transform(table)
data_trans=pd.DataFrame(data_trans,columns=['one','two'])
data_trans
```

```python
imputer=SimpleImputer(missing_values=np.nan,strategy='most_frequent')
imputer
imputer.fit(table)
data_trans=imputer.transform(table)
data_trans=pd.DataFrame(data_trans,columns=['one','two'])
data_trans
```

## Handling the Categorical Data (get_dummies)

```python
import pandas as pd
data={'customer_id':pd.Series([1,2,3,4,5]),
      'loan_type':pd.Series(['Home loan','Personal loan','Education loan','Car loan','Credit loan']),
     'Income':pd.Series(['15k','20k','30k','40k','50k'])}
data=pd.DataFrame(data)
data
```

```python
loan_info=pd.get_dummies(data)
print(loan_info)
```

```python
loan_info=pd.get_dummies(data,dtype=int)
print(loan_info)
```

```python
loan_info=pd.get_dummies(data,prefix_sep='_',drop_first=True,dtype=int)
print(loan_info)
```
