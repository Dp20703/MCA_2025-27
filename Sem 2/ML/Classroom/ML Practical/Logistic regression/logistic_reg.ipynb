The Sample Data
Imagine we want to predict if a student Passes (1) or Fails (0) based on Hours Studied.

Student	Hours Studied(X)  Result(y) 

A	        1	        0
B	        2	        0
C	        4	        1
D	        5	        1



```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# 1. Prepare Data
X = np.array([[1], [2], [4], [5]]) # Hours Studied
y = np.array([0, 0, 1, 1])         # Pass/Fail

# 2. Fit the Model
# Penalty=None allows us to see pure math without regularization
model = LogisticRegression(penalty=None)
model.fit(X, y)

# 3. Extract Mathematical Parameters
intercept = model.intercept_[0]
slope = model.coef_[0][0]

print(f"Mathematical Formula: z = {intercept:.2f} + {slope:.2f} * x")

# 4. Predict for a new value (e.g., 3 hours)
hours = 3
z = intercept + slope * hours
probability = 1 / (1 + np.exp(-z))

print(f"For {hours} hours of study:")
print(f" - Linear score (z): {z:.2f}")
print(f" - Probability of passing: {probability:.2%}")
```

```python
#scratch
import numpy as np
import matplotlib.pyplot as plt

# 1. THE DATASET
# X: Hours Studied, y: 0 (Fail), 1 (Pass)
X = np.array([1, 2, 4, 5])
y = np.array([0, 0, 1, 1])

# 2. HYPERPARAMETERS & INITIALIZATION
learning_rate = 0.2
epochs = 1000
w = 0.0  # Initial weight
b = 0.0  # Initial bias

# Lists to store history for visualization
history = {
    'epoch': [],
    'loss': [],
    'w': [],
    'b': []
}

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def compute_loss(y_true, y_pred):
    # Log Loss / Binary Cross-Entropy
    epsilon = 1e-15 # To prevent log(0)
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    return -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

# 3. TRAINING LOOP (GRADIENT DESCENT)
for i in range(epochs):
    # -- Forward Pass --
    z = w * X + b
    y_pred = sigmoid(z)
    
    # -- Calculate Loss --
    current_loss = compute_loss(y, y_pred)
    
    # -- Backward Pass (Gradients) --
    # Derivative of Loss with respect to w and b
    error = y_pred - y
    dw = np.mean(error * X)
    db = np.mean(error)
    
    # -- Update Parameters --
    w = w - learning_rate * dw
    b = b - learning_rate * db
    
    # Record History
    history['epoch'].append(i)
    history['loss'].append(current_loss)
    history['w'].append(w)
    history['b'].append(b)

    if i % 200 == 0:
        print(f"Epoch {i}: Loss {current_loss:.4f} | w: {w:.3f} | b: {b:.3f}")

# 4. VISUALIZATION
plt.figure(figsize=(16, 5))

# Plot 1: The Loss Curve (Optimization Progress)
plt.subplot(1, 3, 1)
plt.plot(history['epoch'], history['loss'], color='red')
plt.title('Loss Reduction over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Log Loss')
plt.grid(True, alpha=0.3)

# Plot 2: Parameters (w and b) evolution
plt.subplot(1, 3, 2)
plt.plot(history['epoch'], history['w'], label='Weight (w)', color='blue')
plt.plot(history['epoch'], history['b'], label='Bias (b)', color='green')
plt.title('Weight & Bias Evolution')
plt.xlabel('Epoch')
plt.ylabel('Value')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 3: Final Sigmoid Curve Fit
plt.subplot(1, 3, 3)
plt.scatter(X, y, color='black', label='Actual Data', zorder=5)
X_test = np.linspace(0, 6, 100) # Generate 100 points for a smooth curve
y_sigmoid = sigmoid(w * X_test + b)
plt.plot(X_test, y_sigmoid, color='orange', linewidth=3, label='Learned Sigmoid')
plt.axhline(y=0.5, color='gray', linestyle='--', label='Decision Boundary (0.5)')
plt.title('Final Logistic Fit')
plt.xlabel('Hours Studied')
plt.ylabel('Probability of Passing')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Final Equation: P = 1 / (1 + e^-({w:.2f}x + {b:.2f}))")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. DATASET (Many features to 1 output)
# Features: [Hours Studied, Previous Grade %]
X = np.array([
    [1, 50], [2, 45], [4, 70], [5, 85], [1, 80], [5, 40]
])
y = np.array([0, 0, 1, 1, 0, 1]).reshape(-1, 1) # 0=Fail, 1=Pass

# 2. INITIALIZATION
m, n = X.shape # m samples, n features
W = np.zeros((n, 1)) # Weights for each feature
b = 0
lr = 0.01
epochs = 2000
losses = []

# 3. TRAINING
for i in range(epochs):
    # Forward Pass
    z = np.dot(X, W) + b
    y_hat = 1 / (1 + np.exp(-z))
    
    # Loss (Log Loss)
    loss = -np.mean(y * np.log(y_hat + 1e-9) + (1-y) * np.log(1-y_hat + 1e-9))
    losses.append(loss)
    
    # Gradients
    dz = y_hat - y
    dw = (1/m) * np.dot(X.T, dz)
    db = (1/m) * np.sum(dz)
    
    # Update
    W -= lr * dw
    b -= lr * db

# Visualization of Many-to-1
plt.plot(losses)
plt.title("Many-to-1 Loss (Multiple Features)")
plt.xlabel("Epoch"); plt.ylabel("Loss")
plt.show()

print(f"Weights for features: {W.flatten()}")
print(f"Bias: {b}")
```
