## Q.10 Add salt-and-pepper noise to an image and remove the noise using a median filter

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image
img = cv2.imread("image.jpg")

# Convert BGR to RGB for processing/display
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Create a copy of the image
noisy_img = img_rgb.copy()

# Number of noisy pixels
num_pixels = 5000

# Get image height and width
height, width = img_rgb.shape[:2]

# Add salt noise (white pixels)
for i in range(num_pixels // 2):
    x = np.random.randint(0, width)
    y = np.random.randint(0, height)
    noisy_img[y, x] = [255, 255, 255]

# Add pepper noise (black pixels)
for i in range(num_pixels // 2):
    x = np.random.randint(0, width)
    y = np.random.randint(0, height)
    noisy_img[y, x] = [0, 0, 0]

# Remove noise using median filter
denoised_img = cv2.medianBlur(noisy_img, 5)
```

```python
# Display images
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(noisy_img)
plt.title("Salt and Pepper Noise")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(denoised_img)
plt.title("Median Filter Result")
plt.axis("off")

plt.show()
```

## Q.14 Apply Gaussian, median, and bilateral filters to the same noisy image and compare their results

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("image.jpg")

# Convert BGR to RGB for display
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Create a copy to add noise
noisy_img = img_rgb.copy()

# Get image dimensions
height, width = noisy_img.shape[:2]

# Number of noisy pixels
num_pixels = 5000

# Add salt-and-pepper noise
for i in range(num_pixels):

    # Generate random coordinates
    x = np.random.randint(0, width)
    y = np.random.randint(0, height)

    # Add salt or pepper noise
    if i % 2 == 0:
        noisy_img[y, x] = [255, 255, 255]  # Salt (White)
    else:
        noisy_img[y, x] = [0, 0, 0]        # Pepper (Black)


# Apply Gaussian filter
gaussian = cv2.GaussianBlur(noisy_img, (5, 5), 0)

# Apply Median filter
median = cv2.medianBlur(noisy_img, 5)

# Apply Bilateral filter
bilateral = cv2.bilateralFilter(noisy_img, 9, 75, 75)


# Display all images
plt.figure(figsize=(15, 10))

# Original image
plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

# Noisy image
plt.subplot(2, 3, 2)
plt.imshow(noisy_img)
plt.title("Noisy Image")
plt.axis("off")

# Gaussian filter
plt.subplot(2, 3, 3)
plt.imshow(gaussian)
plt.title("Gaussian Filter")
plt.axis("off")

# Median filter
plt.subplot(2, 3, 4)
plt.imshow(median)
plt.title("Median Filter")
plt.axis("off")

# Bilateral filter
plt.subplot(2, 3, 5)
plt.imshow(bilateral)
plt.title("Bilateral Filter")
plt.axis("off")

plt.show()
```

## Q.15 Sharpen an image using a Laplacian-based sharpening technique

```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Read the image
img = cv2.imread("coin.jpg")

# Convert BGR to RGB for display
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Laplacian operator
laplacian = cv2.Laplacian(gray, cv2.CV_64F)

# Convert Laplacian to unsigned 8-bit format
laplacian_abs = cv2.convertScaleAbs(laplacian)

# Sharpen the image
sharpened = cv2.subtract(img_rgb, cv2.cvtColor(laplacian_abs, cv2.COLOR_GRAY2RGB))
```

```python
# Display images
plt.figure(figsize=(12, 4))

# Original image
plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

# Laplacian edges
plt.subplot(1, 3, 2)
plt.imshow(laplacian_abs, cmap="gray")
plt.title("Laplacian Edges")
plt.axis("off")

# Sharpened image
plt.subplot(1, 3, 3)
plt.imshow(sharpened)
plt.title("Sharpened Image")
plt.axis("off")

plt.show()
```

## Q.16 Apply unsharp masking

```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("coin.jpg")

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Create a blurred version
blurred = cv2.GaussianBlur(img_rgb, (5, 5), 0)

# Apply unsharp masking
sharpened = cv2.addWeighted(
    img_rgb, 1.5,
    blurred, -0.5,
    0
)

# Display images
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(blurred)
plt.title("Blurred Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sharpened)
plt.title("Unsharp Masking")
plt.axis("off")

plt.show()
```

## Q.17 Perform contrast stretching on a low-contrast grayscale image

```python
import cv2
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)

# Find minimum and maximum pixel values
min_val = img.min()
max_val = img.max()

# Perform contrast stretching
stretched = ((img - min_val) / (max_val - min_val) * 255).astype("uint8")

# Display original and stretched images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Low-Contrast Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(stretched, cmap="gray")
plt.title("Contrast Stretched Image")
plt.axis("off")

plt.show()
```

## Q.21 Standardize the pixel values of a grayscale image using its mean and standard deviation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image in grayscale
img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)

# Convert image to float
img_float = img.astype(np.float32)

# Calculate mean and standard deviation
mean = img_float.mean()
std = img_float.std()

# Standardize the image
standardized_img = (img_float - mean) / std

# Print values
print("Original Mean:", mean)
print("Original Standard Deviation:", std)

print("Standardized Mean:", standardized_img.mean())
print("Standardized Standard Deviation:", standardized_img.std())

# Display images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(standardized_img, cmap="gray")
plt.title("Standardized Image")
plt.axis("off")

plt.show()
```

## Q.26 Apply erosion and dilation to a binary image and display the results

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)

# Convert grayscale image to binary
_, binary_img = cv2.threshold(
    img,
    127,
    255,
    cv2.THRESH_BINARY
)

# Create kernel
kernel = np.ones((5, 5), np.uint8)

# Apply erosion
erosion = cv2.erode(
    binary_img,
    kernel,
    iterations=1
)

# Apply dilation
dilation = cv2.dilate(
    binary_img,
    kernel,
    iterations=1
)


# Display results
plt.figure(figsize=(15, 5))

# Original grayscale image
plt.subplot(1, 4, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

# Binary image
plt.subplot(1, 4, 2)
plt.imshow(binary_img, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

# Erosion
plt.subplot(1, 4, 3)
plt.imshow(erosion, cmap="gray")
plt.title("Erosion")
plt.axis("off")

# Dilation
plt.subplot(1, 4, 4)
plt.imshow(dilation, cmap="gray")
plt.title("Dilation")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)

# Convert grayscale image to binary
_, binary_img = cv2.threshold(
    img,
    127,
    255,
    cv2.THRESH_BINARY
)

# Create a kernel
kernel = np.ones((5, 5), np.uint8)

# Apply morphological opening
opening = cv2.morphologyEx(
    binary_img,
    cv2.MORPH_OPEN,
    kernel
)

# Apply morphological closing
closing = cv2.morphologyEx(
    binary_img,
    cv2.MORPH_CLOSE,
    kernel
)

# Display results
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(binary_img, cmap="gray")
plt.title("Original Binary Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(opening, cmap="gray")
plt.title("Opening")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(closing, cmap="gray")
plt.title("Closing")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## Q.28 Apply Sobel, Prewitt, Laplacian, and Canny edge detection to the same image and compare the detected edges

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread("coin.jpg", cv2.IMREAD_GRAYSCALE)

# -------------------------
# 1. Sobel Edge Detection
# -------------------------

sobel_x = cv2.Sobel(
    img,
    cv2.CV_64F,
    1,
    0,
    ksize=3
)

sobel_y = cv2.Sobel(
    img,
    cv2.CV_64F,
    0,
    1,
    ksize=3
)

# Combine X and Y edges
sobel = cv2.magnitude(sobel_x, sobel_y)

sobel = cv2.convertScaleAbs(sobel)


# -------------------------
# 2. Prewitt Edge Detection
# -------------------------

prewitt_x_kernel = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
])

prewitt_y_kernel = np.array([
    [-1, -1, -1],
    [ 0,  0,  0],
    [ 1,  1,  1]
])

prewitt_x = cv2.filter2D(
    img,
    cv2.CV_64F,
    prewitt_x_kernel
)

prewitt_y = cv2.filter2D(
    img,
    cv2.CV_64F,
    prewitt_y_kernel
)

# Combine X and Y edges
prewitt = cv2.magnitude(
    prewitt_x,
    prewitt_y
)

prewitt = cv2.convertScaleAbs(prewitt)


# -------------------------
# 3. Laplacian Edge Detection
# -------------------------

laplacian = cv2.Laplacian(
    img,
    cv2.CV_64F
)

laplacian = cv2.convertScaleAbs(laplacian)


# -------------------------
# 4. Canny Edge Detection
# -------------------------

canny = cv2.Canny(
    img,
    100,
    200
)


# -------------------------
# Display Results
# -------------------------

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(sobel, cmap="gray")
plt.title("Sobel Edges")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(prewitt, cmap="gray")
plt.title("Prewitt Edges")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian Edges")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(canny, cmap="gray")
plt.title("Canny Edges")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## Q.29 Detect the edges of an object using Canny edge detection and display the resulting edge image

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("image.jpg")

# Convert BGR to RGB for displaying
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert the image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Gaussian blur to reduce noise
blurred = cv2.GaussianBlur(gray, (5, 5), 0)

# Apply Canny edge detection
edges = cv2.Canny(blurred, 100, 200)

# Display original and edge image
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap="gray")
plt.title("Canny Edge Detection")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## Q.30 Perform background removal using grayscale conversion, thresholding, and a binary mask, then display the extracted foreground object

```python
import cv2
import matplotlib.pyplot as plt

# Read the original image
img = cv2.imread("coin.jpg")

# Convert BGR to RGB for displaying
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Create a binary mask
_, mask = cv2.threshold(
    gray,
    127,
    255,
    cv2.THRESH_BINARY
)

# Extract foreground using the binary mask
foreground = cv2.bitwise_and(
    img,
    img,
    mask=mask
)

# Convert extracted foreground to RGB
foreground_rgb = cv2.cvtColor(
    foreground,
    cv2.COLOR_BGR2RGB
)
```

```python
# Display results
plt.figure(figsize=(15, 5))

# Original image
plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

# Binary mask
plt.subplot(1, 3, 2)
plt.imshow(mask, cmap="gray")
plt.title("Binary Mask")
plt.axis("off")

# Extracted foreground
plt.subplot(1, 3, 3)
plt.imshow(foreground_rgb)
plt.title("Extracted Foreground")
plt.axis("off")

plt.tight_layout()
plt.show()
```
