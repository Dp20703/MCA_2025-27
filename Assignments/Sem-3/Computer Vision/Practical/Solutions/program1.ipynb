```python
import cv2
import numpy as np

# Load image
img = cv2.imread('image.jpg')

# Convert to gray
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Binary threshold
_, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

# Noise removal
kernel = np.ones((3,3), np.uint8)
opening = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)

# Distance transform
dist = cv2.distanceTransform(opening, cv2.DIST_L2, 5)

# Foreground
_, fg = cv2.threshold(dist, 0.5*dist.max(), 255, 0)
fg = np.uint8(fg)

# Marker labeling
_, markers = cv2.connectedComponents(fg)

# Apply watershed
markers = cv2.watershed(img, markers)

# Mark boundaries
img[markers == -1] = [0, 0, 255]

# Show result
cv2.imshow("Result", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Load image
img = cv2.imread('image.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# -------------------------
# 🔹 SOBEL FILTER
# -------------------------

# Sobel X and Y
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

# Convert to absolute
sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)

# Combine both directions
sobel = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

# -------------------------
# 🔹 PREWITT FILTER
# -------------------------

# Prewitt kernels
kernelx = np.array([[1, 0, -1],
                    [1, 0, -1],
                    [1, 0, -1]])

kernely = np.array([[1, 1, 1],
                    [0, 0, 0],
                    [-1, -1, -1]])

# Apply filter
prewitt_x = cv2.filter2D(gray, -1, kernelx)
prewitt_y = cv2.filter2D(gray, -1, kernely)

# Combine
prewitt = cv2.addWeighted(prewitt_x, 0.5, prewitt_y, 0.5, 0)

# -------------------------
# SHOW RESULTS
# -------------------------

cv2.imshow("Original Image", img)
cv2.imshow("Gray Image", gray)

cv2.imshow("Sobel X", sobel_x)
cv2.imshow("Sobel Y", sobel_y)
cv2.imshow("Sobel Combined", sobel)

cv2.imshow("Prewitt X", prewitt_x)
cv2.imshow("Prewitt Y", prewitt_y)
cv2.imshow("Prewitt Combined", prewitt)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
