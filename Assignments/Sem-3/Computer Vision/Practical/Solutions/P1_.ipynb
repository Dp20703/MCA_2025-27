## Q1. Read an image using OpenCV and display its width, height, and number of channels

```python
import matplotlib.pyplot as plt
# %pip install opencv-python
import cv2
```

```python
# read the image
img = cv2.imread("image.jpg")

if img is None:
    print("Image not found")
else:
    print("Image loaded successfully")
```

```python
# Get height, width, and number of channels 
height,width,channels = img.shape
```

```python
# Display information
print("Width:", width)
print("Height:", height)
print("Number of Channels:", channels)
```

```python
# Display image
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

## Q3. Resize an image to 224 x 224 pixels and display the original and resized images

```python
import cv2 
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("image.jpg")

if img is None:
    print("Image not found!")
else:
    print("Image Loaded Successfully")
```

```python
# Resize the image
resized_img = cv2.resize(img, (224, 224))

# Display original and resized images
plt.figure(figsize=(8, 4))

# Original image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Resized image
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(resized_img, cv2.COLOR_BGR2RGB))
plt.title("Resized Image")
plt.axis("off")

plt.show()
```

## Q4. Resize an image to 50% of its original dimensions

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("image.jpg")

# Get original dimensions
height, width = img.shape[:2]

# Calculate 50% dimensions
new_width = int(width * 0.5)
new_height = int(height * 0.5)

# Resize the image
resized_img = cv2.resize(img, (new_width, new_height))

# Display original and resized images
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(resized_img, cv2.COLOR_BGR2RGB))
plt.title("50% Resized")
plt.axis("off")

plt.show()
```

## Q5. Crop a rectangular region from an image using specified starting and ending coordinates

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("image.jpg")

# Starting and ending coordinates
x1, y1 = 100, 50
x2, y2 = 400, 300

# Crop the image
cropped = img[y1:y2, x1:x2]

# Display original image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Display cropped image
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
plt.title("Cropped Image")
plt.axis("off")

plt.show()
```
