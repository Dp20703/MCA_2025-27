## 19. Write the code of Finding Covariance from given data.
X= [2, 4, 6, 8, 10] , Y= [1, 3, 5, 7, 9]

```python
import numpy as np

# Given data
X = [2, 4, 6, 8, 10]
Y = [1, 3, 5, 7, 9]

# Step 1: Convert to numpy arrays
X = np.array(X)
Y = np.array(Y)

# Step 2: Calculate means
mean_X = np.mean(X)
mean_Y = np.mean(Y)

# Step 3: Calculate covariance (manual formula)
cov = np.sum((X - mean_X) * (Y - mean_Y)) / len(X)

# Output
print("Mean of X:", mean_X)
print("Mean of Y:", mean_Y)
print("Covariance:", cov)
```

## 24. Do the following pre- processing techniques on the different 10 data sets.

### (a) Data Cleaning (Handling missing value for rows/columns, Handling Duplicates,Outliers detection and removal)

```python
import pandas as pd 
import numpy as np 
from sklearn.preprocessing import StandardScaler,LabelEncoder ,normalize
```

```python
df = pd.read_csv("sample_dataset.csv")
df.head()
```

```python
df.isnull().sum()
```

###  (a) Data Cleaning 

```python
# 1. Handling Missing Values

df["Age"].fillna(df["Age"].mean(), inplace=True)
df["Salary"].fillna(df["Salary"].median(), inplace=True)
```

```python
# 2. Handling Duplicates
df.drop_duplicates(inplace=True)
```

```python
# 3. Outlier Detection & Removal (IQR Method) 
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1 

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR 

df = df[(df['Salary'] >= lower) & df['Salary'] <= upper ]

print("\nAfter Cleaning:\n", df.head())
```

### (b) Handling Categorical data

```python
le = LabelEncoder()
df["City"] = le.fit_transform(df["City"])

print("\nAfter Encoding:\n")
df.head()
```

### (c) SCALING (Standardization)

```python
scaler = StandardScaler()
df[["Age", "Salary"]] = scaler.fit_transform(df[["Age", "Salary"]])

print("\nAfter Scaling:\n")
df.head()
```

### (d) NORMALIZATION

```python
df[['Age','Salary'] ]= normalize((df[['Age','Salary']]))

print("\nAfter Normalization:\n") 
df.head()
```

# stem and leaf

```python
data = [12, 15, 17, 21, 23, 25, 29]

# Sort data
data.sort()

stem_leaf = {}

for num in data:
    stem = num // 10
    leaf = num % 10
    
    if stem not in stem_leaf:
        stem_leaf[stem] = []
    
    stem_leaf[stem].append(leaf)

# Display
for stem, leaf in stem_leaf.items():
    print(stem, "|", *leaf)
```

## (From Scratch Code)

## 1.Mean 

```python
data = [10, 20, 30, 40, 50]

# Step 1: Sum
total = 0
for x in data:
    total += x

# Step 2: Count
n = len(data)

# Step 3: Mean
mean = total / n

print("Mean:", mean)
```

## 2.Median 

```python
data = [10, 20, 30, 40, 50]

# Step 1: Sort
data.sort()

n = len(data)

# Step 2: Find median
if n % 2 == 0:
    median = (data[n//2 - 1] + data[n//2]) / 2
else:
    median = data[n//2]

print("Median:", median)
```

## 3.Mode 

```python
data = [2, 3, 3, 4, 5]

# Count frequency
freq = {}

for x in data:
    if x in freq:
        freq[x] += 1
    else:
        freq[x] = 1

# Find max frequency
max_freq = max(freq.values())

# Find mode
mode = []

for key, value in freq.items():
    if value == max_freq:
        mode.append(key)

print("Mode:", mode)
```

## 2. Variance 

```python
data = [10, 20, 30, 40, 50]

# Mean
mean = sum(data) / len(data)

# Variance
total = 0
for x in data:
    total += (x - mean) ** 2

variance = total / len(data)

print("Variance:", variance)
```

```python
import math

sd = math.sqrt(variance)

print("Standard Deviation:", sd)
```

# Measures of association between two variables

## 1.Covariance

```python
X = [2, 4, 6, 8, 10]
Y = [1, 3, 5, 7, 9]

# Means
mean_X = sum(X) / len(X)
mean_Y = sum(Y) / len(Y)

# Covariance
cov = 0
for i in range(len(X)):
    cov += (X[i] - mean_X) * (Y[i] - mean_Y)

cov = cov / len(X)

print("Covariance:", cov)
```

## 2.Correlation

```python
import math

X = [2, 4, 6, 8, 10]
Y = [1, 3, 5, 7, 9]

mean_X = sum(X) / len(X)
mean_Y = sum(Y) / len(Y)

# Covariance
cov = 0
for i in range(len(X)):
    cov += (X[i] - mean_X) * (Y[i] - mean_Y)
cov = cov / len(X)

# Standard deviations
var_X = sum((x - mean_X)**2 for x in X) / len(X)
var_Y = sum((y - mean_Y)**2 for y in Y) / len(Y)

sd_X = math.sqrt(var_X)
sd_Y = math.sqrt(var_Y)

# Correlation
r = cov / (sd_X * sd_Y)

print("Correlation:", r)
```

# Measures of distribution

## 1.Skewness

```python
# Skewness ~~ ( Mean - Median ) / Standard Deviation
```

```python
data = [10, 20, 30, 40, 100] 

# Mean 
mean = sum(data) / len(data) 

# Median 
data.sort()

n = len(data)

if n % 2 == 0:
    median = (data[n//2 - 1] + data[n//2]) / 2
else:
    median = data[n//2]

# Standard Deviation 
import math
variance = sum((x - mean)**2 for x in data) / len(data)
sd = math.sqrt(variance)

skewness = ( mean - median ) / sd 
print("Skewness:", skewness)
```

## 2. Kurtosis

### Using Library

```python
from scipy.stats import kurtosis

data = [10, 20, 30, 40, 50]

print("Kurtosis:", kurtosis(data))
```

### From Scratch

```python
import math

# Data
data = [10, 20, 30, 40, 50]

n = len(data)

# Step 1: Mean
mean = sum(data) / n

# Step 2: Variance
variance = sum((x - mean)**2 for x in data) / n

# Step 3: Standard Deviation
sd = math.sqrt(variance)

# Step 4: Kurtosis
kurt = sum((x - mean)**4 for x in data) / (n * (sd**4))

print("Kurtosis:", kurt)
```

```python
excess_kurt = kurt - 3
print("Excess Kurtosis:", excess_kurt)
```
