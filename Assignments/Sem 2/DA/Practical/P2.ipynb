## Q.1 Plot the probability mass function (PMF) for the number of heads from 0 to 10.


```python
# 1 
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 10
p = 0.5

# Possible number of heads
x = range(0, 11)

# PMF values
pmf = binom.pmf(x, n, p)

# Plot
plt.bar(x, pmf)

plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.title("PMF of Number of Heads in 10 Coin Tosses")

plt.xticks(x)
plt.show()
```

```python
# 2 

import matplotlib.pyplot as plt 
from math import comb

n = 10 
p = 0.5
x = list(range(1,11))
prob = [comb(n ,k) * (p ** k ) * ( ( 1- p ) ** (n - k)) for k in x]

plt.stem(x,prob) 

plt.title("PMF of numbers of heads ")
plt.xlabel("Number of heads")
plt.ylabel("Probabiltity")

plt.grid(True)
plt.xticks(x)
plt.show()
```

## Q.2 Calculate and plot the cumulative probability P(X ≤ x) for all x from 0 to 10

```python
import matplotlib.pyplot as plt
from math import comb

n = 10
p = 0.5

# Values of x
x = list(range(0, 11))

# PMF values
pmf = [comb(n, k) * (p**k) * ((1-p)**(n-k)) for k in x]

# Cumulative probabilities
cdf = []
cumulative = 0

for prob in pmf:
    cumulative += prob
    cdf.append(cumulative)

# Plot CDF
plt.plot(x, cdf, marker='o')

plt.xlabel("x (Number of Heads)")
plt.ylabel("P(X ≤ x)")
plt.title("Cumulative Probability Distribution")

plt.grid(True)
plt.xticks(x)

plt.show()
```

## Q.4 Plot the Poisson PMF for x = 0 to 10 for λ = 3.

```python
import matplotlib.pyplot as plt
from math import exp, factorial

# Mean value
lam = 3

# x values from 0 to 10
x = list(range(0, 11))

# Poisson PMF values
pmf = [(exp(-lam) * (lam**k)) / factorial(k) for k in x]

# Plot
plt.stem(x, pmf)

plt.xlabel("Number of Cars (x)")
plt.ylabel("Probability")
plt.title("Poisson PMF (λ = 3)")

plt.grid(True)
plt.xticks(x)

plt.show()
```

## Q.5 Repeat Q4 for λ = 5 and λ = 7. Compare the shape of the distributions.

## Q.6 Plot the normal distribution curve and shade the region between 60 and 80.

## Q.9 Plot the standard normal curve and highlight the critical regions for a 95% confidence level.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate x values
x = np.linspace(-4, 4, 1000)

# Standard normal distribution
y = norm.pdf(x)

# Plot curve
plt.plot(x, y)

# Critical values
z1 = -1.96
z2 = 1.96

# Shade left critical region
x_left = np.linspace(-4, z1, 300)
y_left = norm.pdf(x_left)
plt.fill_between(x_left, y_left, alpha=0.5)

# Shade right critical region
x_right = np.linspace(z2, 4, 300)
y_right = norm.pdf(x_right)
plt.fill_between(x_right, y_right, alpha=0.5)

# Draw critical lines
plt.axvline(z1, linestyle='--')
plt.axvline(z2, linestyle='--')

# Labels and title
plt.xlabel("z")
plt.ylabel("Probability Density")
plt.title("Standard Normal Curve with 95% Confidence Level")

plt.grid(True)

plt.show()
```
