```python
#Quantization
#To reduce the number of gray levels.

import cv2

img = cv2.imread("left.jpg",0)
print(img.shape)
#reduces the number of colors in the image.

quantized = (img//64)*64
print(quantized)

cv2.imshow("Quantized Image", quantized)

cv2.waitKey(0)
cv2.destroyAllWindows()

# original 256 gray leval and now 64 gray level

#860 × 1294 = 1,112,840 pixels
# Each of these 1,112,840 pixels stores 3 intensity values: [B, G, R]
# So a pixel might look like: 
# [155 152 154]
# 155 // 64 = 2
# 2 * 64 = 128
# 152 // 64 = 2
# 2 * 64 = 128
# 154 // 64 = 2
# 2 * 64 = 128
# [128, 128, 128] medium gray.

#Instead of having 256 shades, we want fewer shades.
# 0,64,128,192 now there are 4 gray levels
#256/4=64

```

```python
#Gamma Correction
import cv2
import numpy as np

# Read the image
img = cv2.imread("left.jpg")   # Replace with your image path

# Check if image is loaded
if img is None:
    print("Error: Image not found!")
    exit()

# Gamma value
gamma = 2.0

# Create lookup table
table = []

for i in np.arange(256):
    new_value = ((i / 255.0) ** (1 / gamma)) * 255
    print(f"Original: {i:3} --> Corrected: {int(new_value):1}")
    table.append(new_value)

table = np.array(table).astype("uint8")

# Apply gamma correction
corrected = cv2.LUT(img, table)

# Display images
cv2.imshow("Original", img)
cv2.imshow("Gamma Corrected", corrected)

cv2.waitKey(0)
cv2.destroyAllWindows()

#Gamma controls the brightness of an image using a non-linear transformation.
#It does not simply add or subtract brightness. Instead, it changes how dark and bright pixels are distributed.

# The formula is:
# Output=255×(Input/255)^1/γ

# gamma = 1 → No change
# gamma > 1 → Image becomes brighter
# gamma < 1 → Image becomes darker
```

```python
#Log Transformation
import cv2
import numpy as np

img = cv2.imread("left.jpg")

# Convert to float
img = img.astype(np.float32)

# Scaling constant
c = 255 / np.log(1 + np.max(img))

# Log transformation
log = c * np.log(img + 1)

# Convert back to uint8
log = np.uint8(log)

cv2.imshow("Original", cv2.imread("left.jpg"))
cv2.imshow("Log Transformation", log)

cv2.waitKey(0)
cv2.destroyAllWindows()

#Log transformation is an image enhancement technique used to make dark areas of an image brighter while keeping bright areas from becoming too bright.
# The formula is:s=c×log(1+r)
```

```python
#Power Law
# #Power Law Transformation is an image enhancement technique that changes the brightness of an image using a power (exponent).
# Output=255(Input/255)^γ
import cv2
import numpy as np

# Read the image
img = cv2.imread("left.jpg")   # Replace with your image path

# Check if image is loaded
if img is None:
    print("Error: Image not found!")
    exit()

# Convert image to float to avoid integer issues
img_float = img.astype(np.float32)

# Gamma value
gamma = 0.5

# Power law transformation
power = np.array(
    255 * (img_float / 255) ** gamma,
    dtype=np.uint8
)

# Display images
cv2.imshow("Original", img)
cv2.imshow("Power Law (Gamma = 0.5)", power)

cv2.waitKey(0)
cv2.destroyAllWindows()

# | Gamma (γ) | Effect         |
# | --------- | -------------- |
# | γ = 1     | No change      |
# | γ < 1     | Brighter image |
# | γ > 1     | Darker image   |
```

```python
#SHAPE REPRESENTATION
#Find Contours
import cv2

# Read the image
img = cv2.imread("left.jpg")    # Replace with your image name

# Check if image is loaded
if img is None:
    print("Error: Image not found!")
    exit()

# Convert the image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply binary threshold
_, th = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Find contours
contours, hierarchy = cv2.findContours(
    th,
    cv2.RETR_TREE,
    cv2.CHAIN_APPROX_SIMPLE
)

# Draw all contours on the original image
cv2.drawContours(
    img,
    contours,
    -1,             # Draw all contours
    (0, 255, 0),    # Green color
    2               # Thickness
)

# Display images
cv2.imshow("Original Image", cv2.imread("left.jpg"))
cv2.imshow("Grayscale Image", gray)
cv2.imshow("Threshold Image", th)
cv2.imshow("Contours", img)

# Wait until a key is pressed
cv2.waitKey(0)

# Close all windows
cv2.destroyAllWindows()

#How should I find the objects?"
# RETR_TREE means:
# Find all objects (contours).
# Also remember which object is inside another object.
#     Outer rectangle = Object 1
# Inner rectangle = Object 2

# RETR_TREE remembers:

# Object 2 is inside Object 1.

# cv2.CHAIN_APPROX_SIMPLE
# "How many boundary points should I save
# Without simplification
# OpenCV stores every edge pixel.
# Hundreds of points.

# With CHAIN_APPROX_SIMPLE
# OpenCV stores only the important corner points.
# Only 4 points.

# This saves memory and makes processing faster.


# | Code                      | Simple Meaning                                                          |
# | ------------------------- | ----------------------------------------------------------------------- |
# | `th`                      | Search for objects in this image.                                       |
# | `cv2.RETR_TREE`           | Find all objects and remember which ones are inside others.             |
# | `cv2.CHAIN_APPROX_SIMPLE` | Save only the important boundary points (like corners).                 |
# | `contours`                | The outlines (boundaries) of the objects found.                         |
# | `hierarchy`               | Information about which objects are inside or related to other objects. |
```

```python
#Contour Area
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("light.jpg")

# Convert BGR to RGB (for displaying)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Convert grayscale to binary image
_, th = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Find contours
contours, hierarchy = cv2.findContours(
    th,
    cv2.RETR_TREE,
    cv2.CHAIN_APPROX_SIMPLE
)

# Draw all contours
cv2.drawContours(img_rgb, contours, -1, (255, 0, 0), 2)

# Print contour area
for i, cnt in enumerate(contours):
    area = cv2.contourArea(cnt)
    print(f"Contour {i+1} Area = {area}")

# Show image
plt.imshow(img_rgb)
plt.title("Contours")
plt.axis("off")
plt.show()
```

```python
#Convex Hull
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("left.jpg")

# Make a copy for drawing
output = img.copy()

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Convert to binary image
_, th = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Find contours
contours, hierarchy = cv2.findContours(
    th,
    cv2.RETR_TREE,
    cv2.CHAIN_APPROX_SIMPLE
)

# Draw Convex Hull for each contour
for i, cnt in enumerate(contours):

    # Compute convex hull
    hull = cv2.convexHull(cnt)

    # Draw original contour (Green)
    cv2.drawContours(output, [cnt], -1, (0, 255, 0), 2)

    # Draw convex hull (Blue)
    cv2.drawContours(output, [hull], -1, (255, 0, 0), 2)

    print(f"Contour {i+1}")
    print("Contour Points :", len(cnt))
    print("Hull Points    :", len(hull))
    print("-" * 30)

# Convert BGR to RGB for matplotlib
output = cv2.cvtColor(output, cv2.COLOR_BGR2RGB)

# Show image
plt.figure(figsize=(8,6))
plt.imshow(output)
plt.title("Original Contour (Green) and Convex Hull (Blue)")
plt.axis("off")
plt.show()
```

```python
import cv2
import os
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt

# Dataset folder
dataset = "cat_dog/training_set/"

X = []
y = []

# Class names
classes = ["cats", "dogs"]

# Read both folders using one loop
for label, folder in enumerate(classes):

    folder_path = os.path.join(dataset, folder)

    for file in os.listdir(folder_path):

        img_path = os.path.join(folder_path, file)
        print(img_path)
        img = cv2.imread(img_path)
        print(img)
        # ---------------- Image Preprocessing ----------------
        img = cv2.resize(img, (100, 100))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        img = cv2.GaussianBlur(img, (5,5), 0)
        img = cv2.equalizeHist(img)
        img = cv2.Canny(img, 100, 200)

        kernel = np.ones((3,3), np.uint8)
        img = cv2.dilate(img, kernel, iterations=1)
        img = cv2.erode(img, kernel, iterations=1)

        # Convert image to 1D
        img = img.flatten()

        # Store data
        X.append(img)
        y.append(label)

# Convert to NumPy arrays
X = np.array(X)
y = np.array(y)

print("Total Images:", len(X))

# ---------------- Train KNN ----------------
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, y)

print("Model Trained Successfully")

# ---------------- Predict New Image ----------------
test = cv2.imread("cat_dog/test_set/cats/cat.4001.jpg")

original = cv2.cvtColor(test, cv2.COLOR_BGR2RGB)

test = cv2.resize(test, (100,100))
test = cv2.cvtColor(test, cv2.COLOR_BGR2GRAY)
test = cv2.GaussianBlur(test, (5,5), 0)
test = cv2.equalizeHist(test)
test = cv2.Canny(test, 100, 200)

kernel = np.ones((3,3), np.uint8)
test = cv2.dilate(test, kernel, iterations=1)
test = cv2.erode(test, kernel, iterations=1)

test = test.flatten()

prediction = knn.predict([test])

print("Prediction:", classes[prediction[0]].upper())

plt.imshow(original)
plt.title("Prediction: " + classes[prediction[0]].upper())
plt.axis("off")
plt.show()
```

```python
#Watershed Segmentation
import cv2
import numpy as np

# Read image
img = cv2.imread("apple.jpg")   # Change to your image path

# Check if image loaded
if img is None:
    print("Image not found!")
    exit()

# Make a copy because watershed modifies the image
result = img.copy()

# 1. Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 2. Convert grayscale image to binary
#What is Otsu
#automatically chooses a good threshold value for you.So you don't have to write th=175
_, th = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

# 3. Create a 3x3 kernel
kernel = np.ones((3, 3), np.uint8)

# 4. Remove small noise
#opening = remove small noise.
opening = cv2.morphologyEx(
    th,
    cv2.MORPH_OPEN,
    kernel
)

# 5. Calculate distance from each pixel to the nearest background
#How far is each pixel from the nearest background?
#How should I measure distance?"
#L2 measures the straight distance between A and B.
#L2 means normal straight-line distance.
dist = cv2.distanceTransform(
    opening,
    cv2.DIST_L2,
    5
)

# 6. Find sure foreground
#0.7 keeps a medium/smaller, safer center
#0.9 only center
#0.3 small regions
_, fg = cv2.threshold(
    dist,
    0.7 * dist.max(),
    255,
    0
)

# Convert foreground to uint8
#This just changes the data type into the format OpenCV expects.
#If distance >= 70% of maximum:
#        ↓
#       255 (white)

# Otherwise:
#        ↓
#        0 (black)
fg = np.uint8(fg)

# 7. Find unknown region
unknown = cv2.subtract(opening, fg)

# 8. Give each foreground object a different label
_, markers = cv2.connectedComponents(fg)

# 9. Add 1 to all marker labels
#Before:                        After:
# 0 = background                  1: back
# 1 = object 1                    2:obj1
# 2 = object 2                    3: obj2
markers = markers + 1

# 10. Mark unknown region as 0
#Because in watershed, the number 0 has a special meaning:
#I don't know which object this pixel belongs to. You decide
markers[unknown == 255] = 0

# 11. Apply Watershed
markers = cv2.watershed(result, markers)

# 12. Draw watershed boundaries in red
# -1 stands for boundariuers
result[markers == -1] = [0, 0, 255]

# Display results
cv2.imshow("Original", img)
cv2.imshow("Binary", th)
cv2.imshow("Opening", opening)
cv2.imshow("Sure Foreground", fg)
cv2.imshow("Watershed Result", result)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#K-Means Segmentation

import cv2
import numpy as np

# Read image
img = cv2.imread("left.jpg")

# Check if image loaded
if img is None:
    print("Image not found!")
    exit()

# 1. Convert image into a list of pixels
Z = img.reshape((-1, 3))

# 2. Convert pixel values to float32
Z = np.float32(Z)

# 3. Define stopping criteria
criteria = (
    cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER,
    10,
    1.0
)

# 4. Number of clusters
K = 4

# 5. Apply K-Means
ret, label, center = cv2.kmeans(
    Z,
    K,
    None,
    criteria,
    10,
    cv2.KMEANS_RANDOM_CENTERS
)

# 6. Convert cluster centers back to uint8
center = np.uint8(center)
print(center)

# 7. Replace every pixel with its cluster center
result = center[label.flatten()]
print(result)

# 8. Reshape back to original image shape
result = result.reshape(img.shape)

# Display images
cv2.imshow("Original", img)
cv2.imshow("K-Means Segmentation", result)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#SNAKES (ACTIVE CONTOUR)
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.segmentation import active_contour

# Read image
img = cv2.imread("left.jpg")

# Check if image loaded
if img is None:
    print("Image not found!")
    exit()

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Convert grayscale to float
gray = gray.astype(np.float32) / 255.0

# Create an initial circular snake
center_y = gray.shape[0] // 2
center_x = gray.shape[1] // 2
#Choose the starting radius
#This creates a starting circle.
# height = 860
# width = 1294
# min = 860
# 860 // 4 = 215
radius = min(gray.shape[0], gray.shape[1]) // 4
print(radius)
#theta = np.linspace(0, 2 * np.pi, 200)
#Create 200 points around the circle.
#              π/2
#               ↑
#               |
# π  ←──────────●──────────→  0 / 2π
#               |
#               ↓
#              3π/2
#Use 200 points to draw the circle.
theta = np.linspace(0, 2 * np.pi, 200)

init = np.column_stack([
    center_y + radius * np.sin(theta),
    center_x + radius * np.cos(theta)
])

# Apply Active Contour
snake = active_contour(
    gray,
    init,
    alpha=0.01,
    beta=1.0,
    gamma=0.01
)

# Display result
plt.figure(figsize=(10, 6))

plt.imshow(img_rgb)

# Initial snake
plt.plot(
    init[:, 1],
    init[:, 0],
    '--',
    linewidth=2,
    label="Initial Snake"
)

# Final snake
plt.plot(
    snake[:, 1],
    snake[:, 0],
    '-',
    linewidth=2,
    label="Final Snake"
)

plt.title("Active Contour (Snake) Segmentation")
plt.legend()
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------
# 1. Read image
# --------------------------------
img = cv2.imread("apple.jpg")

if img is None:
    print("Image not found!")
    exit()

# --------------------------------
# 2. Convert BGR to HSV
# --------------------------------
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# --------------------------------
# 3. Create mask for colored apple
# --------------------------------
# White background has very low saturation.
# Apple has higher saturation.

lower = np.array([0, 40, 20])
upper = np.array([179, 255, 255])
#t checks every pixel in the HSV image.
# If the pixel is inside the range:255 white
# If it is outside: 0->black
mask = cv2.inRange(hsv, lower, upper)

# --------------------------------
# 4. Clean the mask
# --------------------------------
kernel = np.ones((5, 5), np.uint8)

mask = cv2.morphologyEx(
    mask,
    cv2.MORPH_OPEN,
    kernel
)

mask = cv2.morphologyEx(
    mask,
    cv2.MORPH_CLOSE,
    kernel
)

# --------------------------------
# 5. Find contours
# --------------------------------
contours, hierarchy = cv2.findContours(
    mask,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_NONE
)

if len(contours) == 0:
    print("No contour found!")
    exit()

# --------------------------------
# 6. Select largest contour
# --------------------------------
cnt = max(contours, key=cv2.contourArea)

# --------------------------------
# 7. Convert contour to complex
#    numbers
# --------------------------------
complex_pts = (
    cnt[:, 0, 0] +
    1j * cnt[:, 0, 1]
)

# --------------------------------
# 8. Fourier Transform
# --------------------------------
fd = np.fft.fft(complex_pts)

print("Fourier Descriptors:")
print(fd)

# --------------------------------
# 9. Draw contour on original
# --------------------------------
result = img.copy()
#Draw all the contours provided."
cv2.drawContours(
    result,
    [cnt],
    -1,
    (0, 255, 0),
    3
)

# --------------------------------
# 10. Convert for matplotlib
# --------------------------------
img_rgb = cv2.cvtColor(
    img,
    cv2.COLOR_BGR2RGB
)

result_rgb = cv2.cvtColor(
    result,
    cv2.COLOR_BGR2RGB
)

# --------------------------------
# 11. Fourier magnitude
# --------------------------------
magnitude = np.abs(fd)

# --------------------------------
# 12. Display results
# --------------------------------
plt.figure(figsize=(12, 8))

# Original
plt.subplot(2, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

# Mask
plt.subplot(2, 2, 2)
plt.imshow(mask, cmap="gray")
plt.title("Apple Mask")
plt.axis("off")

# Apple contour
plt.subplot(2, 2, 3)
plt.imshow(result_rgb)
plt.title("Apple Contour")
plt.axis("off")

# Fourier descriptors
plt.subplot(2, 2, 4)
plt.plot(magnitude)
plt.title("Fourier Descriptors")
plt.xlabel("Frequency")
plt.ylabel("Magnitude")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
from sklearn.datasets import load_iris
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV

# Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# SVM
svm = SVC()

# Parameters
# c Regularization small c strong reg
params = {
    'C': [0.1, 1, 10],
    'kernel': ['linear', 'rbf']
}

# Grid Search
grid = GridSearchCV(svm, params, cv=5)
grid.fit(X, y)

print("Grid Search:", grid.best_params_)

# Random Search
random = RandomizedSearchCV(
    svm,
    params,
    n_iter=3,
    cv=5,
    random_state=42
)

random.fit(X, y)

print("Random Search:", random.best_params_)
```

```python
from sklearn.datasets import load_iris
from xgboost import XGBClassifier
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
#"How much error did my previous model make? Build the next tree to reduce that error."
#So XGBoost focuses on reducing the prediction error using gradients.
#In machine learning, gradient tells us which direction we should move to reduce the error.
#The negative sign tells us:incerease prediciton it need to go up
#Gradient = Prediction - Actual
#Gradient = +3 → prediction needs to go down
# Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# XGBoost
xgb = XGBClassifier(
    eval_metric='mlogloss',
    random_state=42
)

params = {
    'n_estimators': [50, 100, 150],
    'max_depth': [2, 3, 5],
    'learning_rate': [0.01, 0.1, 0.2]
}

# Grid Search
grid = GridSearchCV(xgb, params, cv=5)
grid.fit(X, y)

print("Grid Search:", grid.best_params_)

# Random Search
random = RandomizedSearchCV(
    xgb, params, n_iter=3, cv=5, random_state=42
)
random.fit(X, y)

print("Random Search:", random.best_params_)
```

```python

import cv2
import numpy as np
import sys

# ---- 1. Load image ----
image_path = "apple.jpg"
image = cv2.imread(image_path)

if image is None:
    print(f"Could not load image at {image_path}")
    sys.exit(1)

output = image.copy()

# ---- 2. Convert to grayscale ----
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# ---- 3. Blur to reduce noise ----
blurred = cv2.GaussianBlur(gray, (5, 5), 0)

# ---- 4. Threshold to separate objects from background ----
# Otsu's method automatically picks the best threshold value
_, thresh = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

# ---- 5. Clean up small noise using morphological opening ----
kernel = np.ones((3, 3), np.uint8)
cleaned = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel, iterations=2)

# ---- 6. Find contours (object boundaries) ----
contours, _ = cv2.findContours(cleaned, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# ---- 7. Filter small contours (noise) and draw bounding boxes ----
min_area = 500  # adjust based on your image size
object_count = 0

for cnt in contours:
    area = cv2.contourArea(cnt)
    if area < min_area:
        continue

    object_count += 1
    x, y, w, h = cv2.boundingRect(cnt)

    # Draw bounding box
    cv2.rectangle(output, (x, y), (x + w, y + h), (0, 255, 0), 2)

    # Label each detected object
    cv2.putText(output, f"Object {object_count}", (x, y - 10),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

print(f"Detected {object_count} object(s).")

# ---- 8. Save and show results ----
cv2.imwrite("detected_output.jpg", output)
cv2.imwrite("threshold_mask.jpg", cleaned)

print("Saved: detected_output.jpg (boxes drawn), threshold_mask.jpg (binary mask)")

# Uncomment to display interactively (requires a GUI environment)
cv2.imshow("Detected Objects", output)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Read image
img = cv2.imread("apple.jpg")

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Blur image two different amounts
blur1 = cv2.GaussianBlur(gray, (5, 5), 1)
blur2 = cv2.GaussianBlur(gray, (5, 5), 3)

# Subtract
dog = cv2.subtract(blur1, blur2)

# Display
cv2.imshow("Original", gray)
cv2.imshow("Blur 1", blur1)
cv2.imshow("Blur 2", blur2)
cv2.imshow("DoG", dog)

cv2.waitKey(0)
cv2.destroyAllWindows()

#So subtraction highlights places where the image behaves differently at the two blur levels.
#Imagine a smooth area:Blur 1 = 100
# Blur 2 = 99
# 100 - 99 = 1
# Very small difference.
#  Now imagine an edge:
# Blur 1 = 180
# Blur 2 = 100
# 180 - 100 = 80
# Large difference!
# Something important is happening here — probably an edge or structure.
#     We subtract two differently blurred images to find the details/structures that change between the two scales.

# Less blurred image
#        -
# More blurred image
#        =
# Details affected by blurring


# | Method            | Simple meaning         | Main purpose                             |
# | ----------------- | ---------------------- | ---------------------------------------- |
# | Gaussian Blur     | Smooth the image       | Remove/reduce noise                      |
# | Morphology        | Clean/modify shapes    | Remove noise, fill gaps, connect objects |
# | Sobel             | Find intensity changes | Detect edges + direction                 |
# | Prewitt           | Find intensity changes | Basic edge detection                     |
```

```python
import cv2
import numpy as np

# 1. Read image
img = cv2.imread("apple.jpg")

# 2. Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 3. Create two Gaussian blurred images
blur1 = cv2.GaussianBlur(gray, (5, 5), 1)
blur2 = cv2.GaussianBlur(gray, (5, 5), 3)

# 4. Convert to float
blur1 = blur1.astype(np.float32)
blur2 = blur2.astype(np.float32)

# 5. Difference of Gaussians
dog = blur1 - blur2

# 6. Normalize so that it becomes visible
dog = cv2.normalize(
    dog,
    None,
    0,
    255,
    cv2.NORM_MINMAX
)

# 7. Convert to uint8
dog = dog.astype(np.uint8)

# 8. Display
cv2.imshow("Original", gray)
cv2.imshow("Blur 1", blur1.astype(np.uint8))
cv2.imshow("Blur 2", blur2.astype(np.uint8))
cv2.imshow("DoG", dog)

cv2.waitKey(0)
cv2.destroyAllWindows()

# But HOW does the computer calculate it?
# Step 1: Find the minimum and maximum
# -2   -1    0    1    2
# ↑              ↑
# MIN            MAX
# min = -2
# max = 2
# Step 2: Map the minimum to 0
# -2 → 0
# Step 3: Map the maximum to 255
# 2 → 255
# Step 4: What about values in between?

# OpenCV keeps their relative position.
#     -2 ───── -1 ───── 0 ───── 1 ───── 2
#     0 ───── 64 ───── 128 ───── 191 ───── 255
#     thats the stretching
#     newValue=(oldvlaue-min)/max-min*(newvlaue-newmin)+newmin
#  min = -2
# max = 2

# newMin = 0
# newMax = 255

# Let's calculate -1.
# -1-(-2)/2-(-2)*255
# 64
```

```python
import cv2
import numpy as np

# 1. Read image
image = cv2.imread("apple.jpg")
# 2. Find edges
edges = cv2.Canny(image, 50, 150)

# 3. Find lines
# 1 This is the rho resolution.
# It controls how precisely the distance of a line is measured.
# 1 means we use a resolution of 1 pixel.

# This is the theta (angle) resolution.
# π radians = 180°.
# Therefore
# π / 180 = 1°
# So we're checking possible line angles with about 1-degree steps.

# This is the minimum number of votes needed to consider something a line.
# "The detected line should be at least 50 pixels long."

# If the gap is 10 pixels or less, HoughLinesP can treat the pieces as one line.

# Look at the edge image, search for straight lines at different angles, require enough votes, ignore lines shorter than 50 pixels, and allow gaps up to 10 pixels.
lines = cv2.HoughLinesP(
    edges,
    1,
    np.pi / 180,
    50,
    minLineLength=50,
    maxLineGap=10
)

# 4. Draw the lines
if lines is not None:
    for line in lines:
        x1, y1, x2, y2 = line[0]

        cv2.line(
            image,
            (x1, y1),
            (x2, y2),
            (0, 255, 0),
            2
        )

# 5. Show result
cv2.imshow("Lines", image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# 1. Load image
image = cv2.imread("sqaure.png")

if image is None:
    print("Image not found!")
    exit()

# 2. Find edges
edges = cv2.Canny(image, 50, 150)

# 3. Find lines
lines = cv2.HoughLinesP(
    edges,
    1,
    np.pi / 180,
    50,
    minLineLength=50,
    maxLineGap=10
)

# 4. Draw lines
if lines is not None:
    for line in lines:
        x1, y1, x2, y2 = line[0]

        cv2.line(
            image,
            (x1, y1),
            (x2, y2),
            (0, 255, 0),
            2
        )

# 5. Save the output
cv2.imwrite("output.jpg", image)

# 6. Show the output
cv2.imshow("Detected Lines", image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

image = cv2.imread("sqaure.png")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

gray = np.float32(gray)

corners = cv2.cornerHarris(
    gray,
    2,
    3,
    0.04
)
corners = cv2.dilate(corners, None)
image[corners > 0.01 * corners.max()] = [0, 0, 255]

cv2.imshow("Harris Corners", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Read image
img = cv2.imread("apple.jpg", 0)

# Create Hessian detector
hessian = cv2.xfeatures2d.HarrisLaplaceFeatureDetector_create()

# Detect keypoints
keypoints = hessian.detect(img)

# Draw keypoints
output = cv2.drawKeypoints(
    img, keypoints, None,
    flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
)

# Display
cv2.imshow("Hessian Features", output)
cv2.waitKey(0)
cv2.destroyAllWindows()

print("Number of features:", len(keypoints))
```
