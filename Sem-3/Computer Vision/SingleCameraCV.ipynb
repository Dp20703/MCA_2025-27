```python
import cv2
import matplotlib.pyplot as plt

# 1. Load image
img = cv2.imread("D:/DEPT_OF_COMPUTER_SCIENCE/COMPUTER_VISION/apple.jpg")

# 2. Get dynamic dimensions (height, width)
height, width, _ = img.shape

# 3. Find the exact center point
center_x = width // 2
center_y = height // 2

# 4. Set radius relative to image size (e.g., 25% of the smallest side)
radius = min(height, width) // 4

# 5. Draw a thick GREEN circle (BGR: 0, 255, 0) — pops instantly against a red apple
cv2.circle(img, (center_x, center_y), radius, (0, 255, 0), thickness=8)

# 6. Convert BGR -> RGB for Matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 7. Display
plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.title(f"Image Center: ({center_x}, {center_y})")
plt.axis("on")
plt.show()
```

```python
import cv2
import numpy as np

FILE_NAME = 'D:/DEPT_OF_COMPUTER_SCIENCE/COMPUTER_VISION/apple.jpg'

# Read image from the disk.
img = cv2.imread(FILE_NAME)

# Shape of image in terms of pixels.
(rows, cols) = img.shape[:2]

# getRotationMatrix2D creates a matrix needed 
# for transformation. We want matrix for rotation 
# w.r.t center to 45 degree without scaling.
M = cv2.getRotationMatrix2D((cols / 2, rows / 2), 45, 1)
res = cv2.warpAffine(img, M, (cols, rows))

cv2.imshow("GeeksforGeeks", res)

cv2.waitKey(0)
cv2.destroyAllWindows()

#1   = original size
# 0.5 = half size
# 2   = double size

#Scale controls zoom.
```
